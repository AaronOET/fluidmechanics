Georeferences the Nalluri Moody chart image (x = log₁₀ Re, y = f) and overlays the Colebrook–White curve for a user-specified relative roughness **ε/D**. Enter a Reynolds number **Re** to mark the point on the curve and read off **f**.

All logic lives in `moody_overlay.py`.

**Image credit:** Background Moody chart from *Nalluri & Featherstone's Civil Engineering Hydraulics: Essential Theory with Worked Examples*, 6th Edition, Figure 4.2, p. 92.

```python
%matplotlib inline
import os, urllib.request
from pathlib import Path
from IPython.display import display

# Fetch the helper modules and chart image when running standalone (e.g. on Colab)
BASE_URL = "https://raw.githubusercontent.com/AaronOET/fluidmechanics/main/moodydiagram/"
for f in ["moody_overlay.py", "colebrook_white.py", "bgimgs/Moody_Nalluri.jpg"]:
    if not os.path.exists(f):
        os.makedirs(os.path.dirname(f) or ".", exist_ok=True)
        urllib.request.urlretrieve(BASE_URL + f, f)

try:
    import rasterio
except ImportError:
    %pip install -q rasterio

from moody_overlay import georeference, moody_widget

workspace_dir = Path.cwd()
```

```python
geoRaster = georeference(workspace_dir / "bgimgs" / "Moody_Nalluri.jpg",
                         workspace_dir / "RAS" / "georef2.tif")
```

```python
display(moody_widget(geoRaster, eps_D=0.008, Re=1e5))
```
